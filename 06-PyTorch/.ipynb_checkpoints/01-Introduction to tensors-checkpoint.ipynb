# Introduction to tensors

```python
import torch
```

## Creating tensors

1. Scalar tensor 0-dimension

```python
scalar = torch.tensor(42)
print("Scalar tensor:", scalar)
print("Scalar dimension:", scalar.ndim)
print("Shape of scalar tensor:", scalar.shape)
```

2. Vector tensor 1-dimension

```python
scalar = torch.tensor([1,2,3])
print("Scalar tensor:", scalar)
print("Scalar dimension:", scalar.ndim)
print("Shape of scalar tensor:", scalar.shape)
```

3. Vector tensor 2-dimension

```python
scalar = torch.tensor([[1,2],[3,4]])
print("Scalar tensor:", scalar)
print("Scalar dimension:", scalar.ndim)
print("Shape of scalar tensor:", scalar.shape)
```

4. Vector tensor 3-dimension

```python
scalar = torch.tensor([
                        [[1,2],[3,4]], 
                        [[1,2],[3,4]]
                        ])

print("Scalar tensor:", scalar)
print("Scalar dimension:", scalar.ndim)
print("Shape of scalar tensor:", scalar.shape)
```

# View and reshape operation

```python
import torch
```

```python
tensor = torch.arange(12)
tensor
```

```python
tensor_view = tensor.view(2,6)
tensor_reshape = tensor.reshape(2,6)

print(tensor_view)
print(tensor_view.is_contiguous())
print(tensor_reshape)
print(tensor_reshape.is_contiguous())
```

```python
inferred_tensor = tensor.view(3,-1)
print(inferred_tensor.shape)
print(inferred_tensor.is_contiguous())
print(inferred_tensor)
```

```python
tensor3d = torch.arange(24).reshape(2,3,4)
print(tensor3d)
print(tensor3d.is_contiguous())
```

```python
flattened = tensor3d.view(-1)
print(flattened.is_contiguous())
print(flattened)
```

```python
flattened = tensor3d.reshape(-1)
flattened
```

```python
tensor2d = torch.arange(24).reshape(12,2)
print(tensor2d)
```

```python
transpose_tensor = tensor2d.t()
print(transpose_tensor.is_contiguous())
print(transpose_tensor)
```

```python
# error = transpose_tensor.view(6,4) # error
transpose_reshape = transpose_tensor.reshape(6,4)
transpose_reshape
```

```python
contiguous_tensor = transpose_tensor.contiguous()
contiguous_tensor
```

```python
contiguous_tensor.view(6,4)
```

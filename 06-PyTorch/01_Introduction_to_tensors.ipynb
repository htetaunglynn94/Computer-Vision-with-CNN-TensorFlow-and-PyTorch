# Introduction to tensors

```python
import torch
```

## Creating tensors

**Scalar tensor 0-dimension**

```python
scalar = torch.tensor(42)
print("Scalar tensor:", scalar)
print("Scalar dimension:", scalar.ndim)
print("Shape of scalar tensor:", scalar.shape)
```

**Vector tensor 1-dimension**

```python
vector = torch.tensor([1,2,3])
print("Scalar tensor:", vector)
print("Scalar dimension:", vector.ndim)
print("Shape of scalar tensor:", vector.shape)
```

**Vector tensor 2-dimension**

```python
matrix = torch.tensor([[1,2],[3,4]])
print("Scalar tensor:", matrix)
print("Scalar dimension:", matrix.ndim)
print("Shape of scalar tensor:", matrix.shape)
```

**Vector tensor 3-dimension**

```python
tensor_3d = torch.tensor([
                            [[1,2],[3,4]],
                            [[1,2],[3,4]]
                            ])

print("Scalar tensor:", tensor_3d)
print("Scalar dimension:", tensor_3d.ndim)
print("Shape of scalar tensor:", tensor_3d.shape)
```

**Test GPU acceleration**

```python
if torch.cuda.is_available():
    gpu_tensor = vector.to('cuda')
    print("Tensor on GPU:", gpu_tensor)
else:
    print("CUDA is not abailable on this machine.")
```

**Tensor properties**

```python
print("Data type of matrix:", matrix.dtype)
print("Device of Tensor:", matrix.device)
print("Device of Tensor:", gpu_tensor.device)
print("Is tensor on CUDA:", matrix.is_cuda)
print("Is tensor on CUDA:", gpu_tensor.is_cuda)
print("Size:", matrix.size())
print("Number of elements:", matrix.numel())
```

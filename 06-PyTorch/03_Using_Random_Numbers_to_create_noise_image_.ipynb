# Using Random Numbers to create noise image

```python
import torch
import matplotlib.pyplot as plt
```

```python
random_tensor = torch.rand(5)
print(random_tensor)
print(random_tensor.ndim)
```

```python
random_tensor = torch.rand(2,5,3) # 2 x (5 rows, 3 columns)
print(random_tensor)
print(random_tensor.shape)
print(random_tensor.ndim)
```

```python
torch.manual_seed(22)
random_tensor = torch.rand(5)
print(random_tensor)
```

## Generate random integers

```python
random_integers = torch.randint(low=-10, high=10, size=(5,5))
print(random_integers)
```

```python
# Generate random numbers from normal distribution

random_numbers_normal = torch.normal(mean=0.5, std=2, size=(5,))
print(random_numbers_normal)
```

## Create noise grey-scaled image using PyTorch

```python
uniform_noise = torch.rand(size=(256, 256))
print(uniform_noise.shape)
print(type(uniform_noise))
```

```python
plt.imshow(uniform_noise, 'grey')
plt.axis('off')
plt.title("Uniform noise image (Grey-scale)")
plt.show()
```

## Create noise RGB image using PyTorch

```python
uniform_noise_rgb = torch.rand(size=(256, 256, 3))
print(uniform_noise_rgb.shape)
print(type(uniform_noise_rgb))
```

```python
plt.imshow(uniform_noise_rgb)
plt.axis('off')
plt.title("Uniform noise image (RGB)")
plt.show()
```

```python
print("Shape", uniform_noise_rgb.shape)
print("Data type", uniform_noise_rgb.dtype)
print("Min value", torch.min(uniform_noise_rgb))
print("Max value", torch.max(uniform_noise_rgb))
```

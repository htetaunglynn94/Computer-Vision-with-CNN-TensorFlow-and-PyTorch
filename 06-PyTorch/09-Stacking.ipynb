# Stack operation

````python
import torch
````

## Stack operation

**Create 1D tensors**

````python
tensor_1 = torch.tensor([1,2,3])
tensor_2 = torch.tensor([4,5,6])

print(tensor_1.ndim, tensor_2.size())
````

````python
stacked_tensor_dim0 = torch.stack((tensor_1, tensor_2), dim=0)
print(f"Shape: {stacked_tensor_dim0.shape}")
print(stacked_tensor_dim0)

stacked_tensor_dim1 = torch.stack((tensor_1, tensor_2), dim=1)
print(f"Shape: {stacked_tensor_dim1.shape}")
print(stacked_tensor_dim1)
````

````python
tensor_3 = torch.tensor([7,8,9])

stacked_tensor_dim0_new = torch.stack((tensor_1, tensor_2, tensor_3), dim=0)
print(f"Shape: {stacked_tensor_dim0_new.shape}")
print(stacked_tensor_dim0_new)
````

**Create 2D tensors**

````python
tensor1 = torch.tensor([[1,2], [3,4]])
tensor2 = torch.tensor([[5,6], [7,8]])

print(tensor1)
print(tensor2)
````

````python
# Insert a new dimension at index 0
stack_tensor2d_dim0 = torch.stack((tensor1, tensor2), dim=0)

print(f"Shape: {stack_tensor2d_dim0.shape}")
print(f"Dimension: {stack_tensor2d_dim0.ndim}")
print(stack_tensor2d_dim0)
````

````python
# Insert a new dimension at index 1
stack_tensor2d_dim1 = torch.stack((tensor1, tensor2), dim=1)

print(f"Shape: {stack_tensor2d_dim1.shape}")
print(f"Dimension: {stack_tensor2d_dim1.ndim}")
print(stack_tensor2d_dim1)
````

**NOTE**:  
```text
Before: (2, 2)
After:  (2, 2, 2)
             ↑
       New dimension
       inserted at dim=1

Original shape: (2, 2)
Insert size 2 at dimension 1
Result:         (2, 2, 2)

tensor1                 tensor2
┌───────┐               ┌───────┐
│ 1   2 │               │ 5   6 │
│ 3   4 │               │ 7   8 │
└───────┘               └───────┘
    │                       │
    └───────────┬───────────┘
                │
       torch.stack(dim=1)
                │
                ▼
Output shape: (2, 2, 2)

Block 0              Block 1
┌───────┐            ┌───────┐
│ 1   2 │            │ 3   4 │
│ 5   6 │            │ 7   8 │
└───────┘            └───────┘
```
Where did these values come from?
- [1, 2] is row 0 of tensor1.
- [5, 6] is row 0 of tensor2.  
Therefore, Block 0 combines the first rows of both input tensors.

- [3, 4] is row 1 of tensor1.
- [7, 8] is row 1 of tensor2.  
Therefore, Block 1 combines the second rows of both input tensors.

````python
# Insert a new dimension at index 2
stack_tensor2d_dim2 = torch.stack((tensor1, tensor2), dim=2)

print(f"Shape: {stack_tensor2d_dim2.shape}")
print(f"Dimension: {stack_tensor2d_dim2.ndim}")
print(stack_tensor2d_dim2)
````

## Concatenication operation

````python
# Concatenated along rows
concat_tensors_dim0 = torch.cat((tensor1, tensor2), dim=0)

print(f"Shape: {concat_tensors_dim0.shape}")
print(f"Dimension: {concat_tensors_dim0.ndim}")
print(concat_tensors_dim0)

# NOTE: no new dimension is added
````

````python
# Concatenated along columns
concat_tensors_dim1 = torch.cat((tensor1, tensor2), dim=1)

print(f"Shape: {concat_tensors_dim1.shape}")
print(f"Dimension: {concat_tensors_dim1.ndim}")
print(concat_tensors_dim1)

# NOTE: no new dimension is added
````

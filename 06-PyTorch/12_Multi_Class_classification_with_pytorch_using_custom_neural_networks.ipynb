# Multi Class classification with pytorch using custom neural networks

**Dataset URL**: [LINK](https://archive.ics.uci.edu/dataset/53/iris)  

We will use this dataset from [Scikit-learn](https://scikit-learn.org/dev/modules/generated/sklearn.datasets.load_iris.html#sklearn.datasets.load_iris)


````python
import numpy as np
import pandas as pd

# Import scikit-learn libraries
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split as TTS
from sklearn.preprocessing import StandardScaler, LabelEncoder

# Import PyTorch libraries
import torch
import torch.nn as nn
import torch.optim as optim
````

````python
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['target'] = iris.target

df.head()
````

````python
X = df.drop(columns=['target'])
y = df['target']
````

````python
# Split train and test data with stratification
X_train, X_test, y_train, y_test = TTS(X, y, test_size=0.2, random_state=42, stratify=y)
np.unique(y_train, return_counts=True), np.unique(y_test, return_counts=True)
````

````python
# Preprocessing stage
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
````

````python
# Convert numpy array to tensor array
X_train_tensor = torch.tensor(X_train, dtype=torch.float32)
X_test_tensor = torch.tensor(X_test, dtype=torch.float32)
y_train_tensor = torch.tensor(y_train.values, dtype=torch.long) # convert values into 64-bit integers (int64)
y_test_tensor = torch.tensor(y_test.values, dtype=torch.long)   # convert values into 64-bit integers (int64)
````

````python
# Create classifier modle

class IrisClassifier(nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim):
        super(IrisClassifier, self).__init__()

        self.network = nn.Sequential(nn.Linear(input_dim, hidden_dim),
                                     nn.ReLU(),
                                     nn.Linear(hidden_dim, hidden_dim),
                                     nn.ReLU(),
                                     nn.Linear(hidden_dim, output_dim))

    def forward(self, x):
        return self.network(x)
````

````python
class IrisClassifier(nn.Module):
  def __init__(self, input_dim, hidden_dim,output_dim):
    super(IrisClassifier, self).__init__()
    self.network = nn.Sequential(
        nn.Linear(input_dim,hidden_dim),
        nn.ReLU(),
        nn.Linear(hidden_dim,hidden_dim),
        nn.ReLU(),
        nn.Linear(hidden_dim,output_dim),
    )

  def forward(self, x):
    return self.network(x)
````

````python
input_dim = X_train.shape[1]
hidden_dim = 16
output_dim = 3

model = IrisClassifier(input_dim, hidden_dim, output_dim)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(params=model.parameters(), lr=0.01)
````

````python
# Set the total number of training iterations
epochs = 500

# Repeat the training process for the specified number of epochs
for epoch in range(epochs):

    # Set the model to training mode
    model.train()

    # Clear gradients from the previous training step
    optimizer.zero_grad()

    # Pass the training features through the model to get predictions
    pred = model(X_train_tensor)

    # Calculate the loss by comparing predictions with the true target values
    loss = criterion(pred, y_train_tensor) # must be y_train_tensor

    # Calculate gradients of the loss with respect to the model's parameters
    loss.backward()

    # Update the model's parameters using the calculated gradients
    optimizer.step()

    # Print the loss every 10 epochs
    if (epoch + 1) % 10 == 0:

        # Display the current epoch and loss value
        print(f"Epochs: {epoch + 1}/[{epochs}] | Loss: {loss.item():.3f}")
````

````python
# Train the model

epochs = 500
for epoch in range(epochs):
  model.train()
  optimizer.zero_grad()

  predictions = model(X_train_tensor)
  loss = criterion(predictions, y_train_tensor)

  loss.backward()
  optimizer.step()

  if (epoch+1) % 10 == 0:
    print(f"Epoch [{epoch+1}/{epochs}], loss : {loss.item():.4f}")
````

````python
# Set the model to evaluation mode
# This disables training-specific behavior such as Dropout
model.eval()

# Disable gradient calculation because we are only evaluating the model
with torch.no_grad():

    # Pass the test features through the model to get prediction scores
    y_pred = model(X_test_tensor)

    # Select the class with the highest score for each sample
    # dim=1 means selecting across the class dimension
    y_pred_labels = torch.argmax(y_pred, dim=1)

    # Compare predicted labels with the true test labels
    # Sum the correct predictions and divide by the total number of test samples
    accuracy = (y_pred_labels == y_test_tensor).sum().item() / y_test_tensor.size(0)

    # Print the test accuracy to 4 decimal places
    print(f"Test accuracy : {accuracy:.4f}")
````

````python
# Define a function to predict the Iris species from its four measurements
def predict_iris(sepal_length, sepal_width, petal_length, petal_width):

    # Create a 2D NumPy array containing the input measurements
    # The extra pair of brackets represents one sample with four features
    input_data = np.array([[sepal_length, sepal_width, petal_length, petal_width]])

    # Scale the input measurements using the scaler fitted during training
    input_data = scaler.transform(input_data)

    # Convert the scaled NumPy array into a PyTorch tensor
    # Use float32 because the model expects floating-point input
    input_tensor = torch.tensor(input_data, dtype=torch.float32)

    # Set the model to evaluation mode
    model.eval()

    # Disable gradient calculation because this is inference, not training
    with torch.no_grad():

        # Pass the input tensor through the model to get class scores
        prediction = model(input_tensor)

        # Select the index of the class with the highest score
        # .item() converts the single-element tensor into a Python number
        predicted_class = torch.argmax(prediction, dim=1).item()

    # Convert the predicted class index into the corresponding Iris species name
    return iris.target_names[predicted_class]
````

````python
predicted_class = predict_iris(5.1, 3.5, 1.4, 0.2)
print(f"Predicted species : {predicted_class}")
````

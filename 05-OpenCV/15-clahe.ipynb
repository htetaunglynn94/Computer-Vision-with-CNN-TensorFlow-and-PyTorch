```python
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("./data/giraffe-Kenya.png")
img_g = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img_g, 'gray')
```

### Perform on GRAY-SCALE image

```python
hist_equalized_img = cv2.equalizeHist(img_g)
plt.imshow(hist_equalized_img, 'gray')
```

```python
clahe = cv2.createCLAHE(clipLimit=2.0, # controls the clipping of histogram peaks to limit contrast amplification
                        tileGridSize=(4,4)) # divides the image into a grid of 4 columns × 4 rows of tiles  
clahe_equalized_img = clahe.apply(img_g)
plt.imshow(clahe_equalized_img, 'gray')
```

```python
plt.figure(figsize=(12,8))

plt.subplot(1,3,1)
plt.imshow(img_g, cmap='grey')
plt.title("Original image")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(hist_equalized_img, cmap='grey')
plt.title("Histogram equalized image")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(clahe_equalized_img, cmap='grey')
plt.title("Clahe equalized image")
plt.axis('off')

plt.tight_layout()
plt.show()
```

### Perform on COLOR image

```python
yuv_img = cv2.cvtColor(img, cv2.COLOR_BGR2YUV)
yuv_img[:,:,0] = clahe.apply(yuv_img[:,:,0])
clahed_img_rgb = cv2.cvtColor(yuv_img, cv2.COLOR_YUV2RGB)

plt.figure(figsize=(12,8))

plt.subplot(1,2,1)
plt.imshow(img_rgb)
plt.title("Original RGB image")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(clahed_img_rgb)
plt.title("Clahe equalized RGB image")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('data/giraffe-Kenya.png')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
```

## Image resizing

```python
print(img_rgb.shape)
```

```python
new_height, new_width = 400, 400
resized_img = cv2.resize(src=img_rgb, dsize=(new_width, new_height))
plt.imshow(resized_img)
```

# Image Scaling

```python
scaling_factor = 0.5
s_width = int(img_rgb.shape[1] * scaling_factor)
s_height = int(img_rgb.shape[0] * scaling_factor)
scaled_image = cv2.resize(src=img_rgb, dsize=(s_width,s_height), interpolation=cv2.INTER_AREA)

# scaled_images
plt.imshow(scaled_image)
plt.title("Scaled image")
plt.show()
```

## Interpolation

```python
def show_image(images):
    fig, ax = plt.subplots(1,6, figsize=(20,5))

    for i, (title, image) in enumerate(images.items(), start=0):
        ax[i].imshow(image)
        ax[i].set_title(title)
        ax[i].axis('off')

    plt.suptitle('IMAGE INTERPOLATION METHODS')
    plt.tight_layout()
    plt.show()    
```

```python
# Downscaling original image
resized_img_rgb = cv2.resize(src=img_rgb, dsize=(100,100))
# print(resized_img_rgb.shape)

# Upscaling the downscaled image
resize_width = 600
resize_height = 600

# Interploation methods
methods = ['cv2.INTER_NEAREST','cv2.INTER_LINEAR','cv2.INTER_CUBIC','cv2.INTER_AREA','cv2.INTER_LANCZOS4']
resized_images = dict()
resized_images['Original'] = resized_img_rgb

for method in methods:
    alg = eval(method)
    resized_images[method] = cv2.resize(src=resized_img_rgb, dsize=(resize_width,resize_height), interpolation=alg)

# Visualize resized images simultaneously
show_image(resized_images)
```

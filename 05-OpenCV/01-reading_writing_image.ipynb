```python
import cv2
import matplotlib.pyplot as plt
```

```python
# Read the image
image = cv2.imread("data/butterfly.jpg")
img = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
# Check image is loaded or not
if image is None:
    print("Error: could not loaded the image.")
else:
    print("Image is loaded successfully")
```

```python
# Display the image
plt.imshow(img)
```

```python
# Write the image
image_new = image + 30
cv2.imwrite("data/output.jpg", image_new)
```

```python
import cv2
```

```python
# Load the video
video = cv2.VideoCapture("data/mountain.mp4")
```

```python
# Display the video
while True:
    ret, frame = video.read()
    if not ret: # ret == False
        break
    cv2.imshow("Video Frame", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# T0 ---> f1
# wait 1 ms
# # T1 ---> f2

video.release()
cv2.destroyAllWindows()
```

```python
#####################
### Write a video ###
#####################

# Read input video
video = cv2.VideoCapture("data/mountain.mp4")

# Create a VideoWriter object
fourcc = cv2.VideoWriter_fourcc(*"mp4v")

# Get video properties
fps = video.get(cv2.CAP_PROP_FPS)
width = int(video.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(video.get(cv2.CAP_PROP_FRAME_HEIGHT))

out = cv2.VideoWriter(filename="data/output.mp4", 
                      fourcc=fourcc, 
                      fps=fps, 
                      frameSize=(width, height))

while True:
    ret, frame = video.read()
    if not ret:
        break
    out.write(frame)

out.release()
video.release()
cv2.destroyAllWindows()
```

```python
#####################################################
### Write a video frame with updated video frames ###
#####################################################

# Read input video
video = cv2.VideoCapture("data/mountain.mp4")

# Create a VideoWriter object
fourcc = cv2.VideoWriter_fourcc(*"mp4v")

# Get video properties
fps = video.get(cv2.CAP_PROP_FPS)
width = int(video.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(video.get(cv2.CAP_PROP_FRAME_HEIGHT))

out = cv2.VideoWriter(filename="data/output_updated.mp4", 
                      fourcc=fourcc, 
                      fps=fps, 
                      frameSize=(width, height))

while True:
    ret, frame = video.read()
    if not ret:
        break
    frame = frame + 4
    out.write(frame)

out.release()
video.release()
cv2.destroyAllWindows()
```

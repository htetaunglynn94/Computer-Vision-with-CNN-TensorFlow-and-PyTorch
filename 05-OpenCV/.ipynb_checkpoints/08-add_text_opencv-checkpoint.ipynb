```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
height, width = 500, 700
violet = (153,52,255)
canvas = np.full(shape=(height, width, 3), fill_value=violet, dtype=np.uint8)
plt.imshow(canvas)
```

```python
img = canvas.copy()                     # Copy image
text = "Hello OpenCV!"                  # Text to display
org = (100, 250)                        # Bottom-left corner of the text (x, y)
fontFace = cv2.FONT_HERSHEY_COMPLEX     # Font style
fontScale = 1.5                         # Font size multiplier
color = (255, 255, 0)                   # Text color (BGR)
thickness = 2                           # Text thickness in pixels
lineType = cv2.LINE_AA                  # Anti-aliased lines for smoother text

cv2.putText(img, text, org, fontFace, fontScale, color, thickness, lineType)
plt.imshow(img)
```

```python
img = canvas.copy()                     # Copy image
fontScale = 1.5                         # Font size multiplier
color = (255, 255, 0)                   # Text color (BGR)
thickness = 2                           # Text thickness in pixels
lineType = cv2.LINE_AA                  # Anti-aliased lines for smoother text

fonts = {'Simplex': cv2.FONT_HERSHEY_SIMPLEX,
         'Complex': cv2.FONT_HERSHEY_COMPLEX,
         'Script Simplex': cv2.FONT_HERSHEY_SCRIPT_SIMPLEX,
         'Script Complex': cv2.FONT_HERSHEY_SCRIPT_COMPLEX,
         'Complex Small': cv2.FONT_HERSHEY_COMPLEX_SMALL}     

y_offset = 150
for i, (text, fontFace) in enumerate(fonts.items()):
    org = (50, y_offset)                # Bottom-left corner of the text (x, y)
    cv2.putText(img, text, org, fontFace, fontScale, color, thickness, lineType)
    y_offset += 50

plt.imshow(img)
```

## Font size & thickness

```python
img = canvas.copy()                     # Copy image
text = "FONT 2"                         # Text to display
fontScale = 1.5                         # Font size multiplier
color = (255, 255, 0)                   # Text color (BGR)
fontFace = cv2.FONT_HERSHEY_COMPLEX     # Font style
thickness = 2                           # Text thickness in pixels
lineType = cv2.LINE_AA                  # Anti-aliased lines for smoother text

styles = [(1,2), (2,2), (1,6), (2,4)]

y_offset = 150
for fontScale, thickness in styles:
    org = (50, y_offset)                # Bottom-left corner of the text (x, y)
    cv2.putText(img, text, org, fontFace, fontScale, color, thickness, lineType)
    y_offset += 50

plt.imshow(img)
```

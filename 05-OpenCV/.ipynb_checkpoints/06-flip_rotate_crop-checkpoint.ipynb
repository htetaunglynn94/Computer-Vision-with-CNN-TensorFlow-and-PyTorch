```python
import cv2
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
img = cv2.imread('./data/giraffe-Kenya.png')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
```

# Flip

```python
img_c = img.copy()
h = cv2.cvtColor(img_c, cv2.COLOR_BGR2RGB)
plt.imshow(h)
```

```python
# 1 = horizontal | 0 = vertical | -1 = both
flips = {'Original': None,
         'Horizontally flipped': 1, 
         'Vertically flipped': 0, 
         'Flipped both axes': -1}

fig, axs = plt.subplots(2, 2, figsize=(6, 8))

for ax, (title, code) in zip(axs.flat, flips.items()):
    
    image = img_rgb if code is None else cv2.cvtColor(
                cv2.flip(img, flips[title]), cv2.COLOR_BGR2RGB)
    
    ax.imshow(image)
    ax.set_title(title)
    ax.axis('off')

plt.suptitle("Flipping image")
plt.tight_layout()
plt.show()
```

# Rotate

```python
rotations = {'Original': None, 
             'Clockwise (90$^\circ$)': 'cv2.ROTATE_90_CLOCKWISE',
             'Clockwise (180$^\circ$)': 'cv2.ROTATE_180',
             'Counterclockwise (90$^\circ$)': 'cv2.ROTATE_90_COUNTERCLOCKWISE'}

fig, axs = plt.subplots(2, 2, figsize=(8,6))

for ax, (title, code) in zip(axs.flat, rotations.items()):

    image = img_rgb if code is None else cv2.rotate(img_rgb, eval(code))

    ax.imshow(image)
    ax.set_title(title)
    ax.axis('off')

plt.suptitle("Rotating image")
plt.tight_layout()
plt.show()
```

# Cropping and image

```python
print(img_rgb.shape)
plt.imshow(img_rgb)
```

```python
x_topleft, y_topleft   = 350, 300
x_botright, y_botright = 800, 800

cropped_img = img_rgb[x_topleft: x_botright, y_topleft:y_botright]

plt.subplot(1,2,1)
plt.imshow(img_rgb)
plt.title("Original image")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(cropped_img)
plt.title("Cropped image")
plt.axis('off')

plt.tight_layout()
plt.show()
```

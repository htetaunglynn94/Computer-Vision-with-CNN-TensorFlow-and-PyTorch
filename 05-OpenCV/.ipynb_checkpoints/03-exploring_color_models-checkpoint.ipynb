```python
import cv2
import matplotlib.pyplot as plt
```

```python
bgr_img = cv2.imread('data/mountain.jpg')
plt.imshow(bgr_img)
```

```python
# Convert to grayscale
gray_img = cv2.cvtColor(bgr_img, cv2.COLOR_BGR2GRAY)
cv2.imwrite('data/gray_mountain.jpg', gray_img)
plt.imshow(gray_img, 'gray')
```

```python
# Convert to HSV
hsv_img = cv2.cvtColor(bgr_img, cv2.COLOR_BGR2HSV)
cv2.imwrite('data/hsv_mountain.jpg', hsv_img)
plt.imshow(hsv_img)
```

```python
# Convert to LAB
lab_img = cv2.cvtColor(bgr_img, cv2.COLOR_BGR2LAB)
cv2.imwrite('data/lab_mountain.jpg', hsv_img)
plt.imshow(lab_img)
```

```python
# Convert to RGB
rgb_img = cv2.cvtColor(bgr_img, cv2.COLOR_BGR2RGB)
cv2.imwrite('data/rgb_mountain.jpg', rgb_img)
plt.imshow(rgb_img)
```

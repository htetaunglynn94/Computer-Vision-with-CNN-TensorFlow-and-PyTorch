```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Create an empty canvas
height, width = 500, 500
blue = (0,0,255) # RGB channel
canvas = np.full((height, width, 3), blue, dtype=np.uint8)
plt.imshow(canvas)
```

## Draw to a line

```python
img = canvas.copy()
cv2.line(img=img, pt1=(50,50), pt2=(400, 450), color=(255,255,255), thickness=5)
plt.imshow(img)
```

## Draw a rectangle

```python
cv2.rectangle(img=img, pt1=(100,150), pt2=(200, 250), color=(0,255,255), thickness=-1)
plt.imshow(img)
```

# Draw a circle

```python
cv2.circle(img=img, center=(250,250), radius=100, color=(255,0,0), thickness=5)
plt.imshow(img)
```

## Draw an Ellipse

```python
cv2.ellipse(img=img, center=(150,350), axes=(100,50), angle=45, startAngle=0, 
            endAngle=360, color=(0,255,0), thickness=5)
plt.imshow(img)
```

## Draw a polygon

```python
img_cp = canvas.copy()
```

```python
points = np.array([[100,400], [400,400], [250,100]], dtype=np.int32) # 2D | shape: (3,2)

# Here, no need reshaping
# new_point = points.reshape((-1,1,2)) # 1D | shape: (3,1,2)

cv2.polylines(img=img_cp, pts=[points], isClosed=True, color=(255,0,255), thickness=3)
cv2.polylines(img=img_cp, pts=[points], isClosed=False, color=(0,255,255), thickness=3)

plt.imshow(img_cp)
```

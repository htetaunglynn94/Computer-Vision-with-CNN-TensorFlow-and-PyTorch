```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# RGB
img = cv2.imread('./data/giraffe-Kenya.png')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
```

```python
b_ch, g_ch, r_ch = cv2.split(img)  # Split RGB channels
b_ch.shape, g_ch.shape, r_ch.shape
```

```python
channels = {'blue': b_ch, 
            'green': g_ch, 
            'red': r_ch}

for color, channel in channels.items():

    images, channels = [channel], [0]
    mask, histSize, ranges = None, [256], [0,256]

    '''
    Ranging from 0 to 256 means data type unsigned integer 8 bits (uint8) has 2^8 having total 256 values.
    '''

    hist = cv2.calcHist(images, channels, mask, histSize, ranges)
    plt.plot(hist, color=color)
```

```python
img_hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
plt.imshow(img_hsv)
```

```python
h_ch, s_ch, v_ch = cv2.split(img_hsv)
h_ch.shape, s_ch.shape, v_ch.shape
```

```python
fig, axes = plt.subplots(1,3, figsize=(15, 5))

h_hist = cv2.calcHist([h_ch], [0], None, [180], [0, 180])
s_hist = cv2.calcHist([s_ch], [0], None, [256], [0, 256])
v_hist = cv2.calcHist([v_ch], [0], None, [256], [0, 256])

plt.subplot(1,3,1)
plt.plot(h_hist, color='orange')
plt.title("Hue Histogram")
plt.xlabel("Bins")
plt.ylabel("Frequency")

plt.subplot(1,3,2)
plt.plot(s_hist, color='green')
plt.title("Saturation Histogram")
plt.xlabel("Bins")
plt.ylabel("")

plt.subplot(1,3,3)
plt.plot(v_hist, color='blue')
plt.title("Value Histogram")
plt.xlabel("Bins")
plt.ylabel("")

plt.tight_layout()
plt.show()
```

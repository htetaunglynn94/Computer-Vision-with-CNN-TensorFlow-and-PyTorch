```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
img = cv2.imread("./data/giraffe-Kenya.png", 0)
plt.imshow(img, 'grey')
```

## Average blur

```python
kernel = 15
average_blur =  cv2.blur(src=img, ksize=(kernel, kernel))
plt.imshow(average_blur, 'grey')
```

## Gaussian blur

```python
kernel = 15
gaussian_blur = cv2.GaussianBlur(src=img, ksize=(kernel,kernel), sigmaX=0)
plt.imshow(gaussian_blur, 'grey')
```

## Median blur

```python
median_blur = cv2.medianBlur(img, 11)

plt.imshow(median_blur, cmap='grey')
```

```python
plt.figure(figsize=(15,7))

plt.subplot(1,4,1)
plt.imshow(img, 'grey')
plt.title("Original image")
plt.axis('off')

plt.subplot(1,4,2)
plt.imshow(average_blur, 'grey')
plt.title("Average blurred image")
plt.axis('off')

plt.subplot(1,4,3)
plt.imshow(gaussian_blur, 'grey')
plt.title("Gaussian blurred image")
plt.axis('off')

plt.subplot(1,4,4)
plt.imshow(median_blur, 'grey')
plt.title("Median blurred image")
plt.axis('off')

plt.tight_layout()
plt.suptitle("Image Blurring Methods (Average, Gaussian and Median)", weight='bold', size=12)
plt.show()
```

```python
noise = cv2.imread('./data/salt_and_pepper.png', 0)

med_blur = cv2.medianBlur(noise, 9)

plt.figure(figsize=(12,7))
plt.subplot(1,2,1)
plt.imshow(noise, 'grey')
plt.title("Original image")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(med_blur, 'grey')
plt.title("Image after removing noise")
plt.axis('off')

plt.suptitle("Apply median blur on image with salt-and-pepper noise", size=12, weight='bold')
plt.tight_layout()
plt.savefig("median_blur.png")
plt.show()
```

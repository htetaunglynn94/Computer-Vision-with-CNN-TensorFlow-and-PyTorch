Place it inside `./models`.

```python
# Load the YuNet face detection model
model_path = './models/face_detection_yunet_2026may.onnx'

# Create YuNet face detector
detector = cv2.FaceDetectorYN.create(model=model_path, # YuNet face detection ONNX model file path
                                     config="",  # empty configuration file path
                                     input_size=(650, 650)) # initial input image size

# Open Webcam
cam = cv2.VideoCapture(0)

if not cam.isOpened():
    raise RuntimeError("Cannot open webcam.")

while True:
    ret, frame = cam.read()

    if not ret:
        print("Fail to capture frame")
        break

    # Resize the detector's input size to match the webcam frame
    ## Get the height and width of the webcam frame
    height, width = frame.shape[:2] # shape of frame: (480, 640, 3)
    ## Update the face detector's input size to match the webcam frame
    detector.setInputSize((width, height))

    # Detect faces
    retval , faces = detector.detect(frame)

    if faces is not None:
            for face in faces:
                 # Takes only the first 4 values that we require
                x, y, w, h = face[:4].astype(int)

                # Draw rectangle
                cv2.rectangle(img=frame, 
                              pt1=(x-20,y-20), 
                              pt2=(x+w+20, y+h+20), 
                              color=(255, 0, 0), 
                              thickness=2)

                # Add custom text
                cv2.putText(img=frame, 
                            text="Face Detected", 
                            org=(x-20, y-30), 
                            fontFace=cv2.FONT_HERSHEY_SIMPLEX, 
                            fontScale=0.9, 
                            color=(0,255,0), 
                            thickness=2, 
                            lineType=cv2.LINE_4)
                
    # frame_gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    # cv2.imshow("Face detector", frame_gray)

    cv2.imshow("Face detector", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
    
cam.release()
cv2.destroyAllWindows()
```

`detector.detect(frame)` returns two output values:
1. **retval** — a result/status value.
2. **faces** — the detected face information, including bounding boxes, confidence scores, and facial landmarks.


`faces` include 15 values for each one, which represents:

| Index | Value | Meaning |
|---|---|---|
| `0` | `x` | X-coordinate of the bounding box's top-left corner |
| `1` | `y` | Y-coordinate of the bounding box's top-left corner |
| `2` | `w` | Width of the bounding box |
| `3` | `h` | Height of the bounding box |
| `4–5` | `right_eye` | X and Y coordinates of the right eye |
| `6–7` | `left_eye` | X and Y coordinates of the left eye |
| `8–9` | `nose_tip` | X and Y coordinates of the nose |
| `10–11` | `right_mouth_corner` | X and Y coordinates of the right mouth corner |
| `12–13` | `left_mouth_corner` | X and Y coordinates of the left mouth corner |
| `14` | `confidence` | Confidence score for the face detection |

Here, we need only `x`,`y`,`w` and `h`.

### Face, eyes, nose tip and mouth detector

```python
# Load the YuNet face detection model
model_path = './models/face_detection_yunet_2026may.onnx'

# Create YuNet face detector
detector = cv2.FaceDetectorYN.create(model=model_path, # YuNet face detection ONNX model file path
                                     config="",  # empty configuration file path
                                     input_size=(650, 650)) # initial input image size

# Open Webcam
cam = cv2.VideoCapture(0)

if not cam.isOpened():
    raise RuntimeError("Cannot open webcam.")

while True:
    ret, frame = cam.read()

    if not ret:
        print("Fail to capture frame")
        break

    # Resize the detector's input size to match the webcam frame
    ## Get the height and width of the webcam frame
    height, width = frame.shape[:2] # shape of frame: (480, 640, 3)
    ## Update the face detector's input size to match the webcam frame
    detector.setInputSize((width, height))

    # Detect faces
    retval , faces = detector.detect(frame)

    if faces is not None:
            for face in faces:
                # Takes only the first 4 values that we require
                x, y, w, h = face[:4].astype(int)

                # Draw rectangle
                cv2.rectangle(img=frame, 
                              pt1=(x-20,y-20), 
                              pt2=(x+w+20, y+h+20), 
                              color=(255, 0, 0), 
                              thickness=2)

                # Add custom text 1
                cv2.putText(img=frame, 
                            text="Face Detected", 
                            org=(x-20, y-30), 
                            fontFace=cv2.FONT_HERSHEY_SIMPLEX, 
                            fontScale=0.9, 
                            color=(0,255,0), 
                            thickness=2, 
                            lineType=cv2.LINE_4)

                # Add custom text 2
                cv2.putText(img=frame, 
                            text="FACE, EYES, NOSE TIP AND MOUTH DETECTOR", 
                            org=(40, 50), 
                            fontFace=cv2.FONT_HERSHEY_SIMPLEX, 
                            fontScale=0.9, 
                            color=(0,0,255), 
                            thickness=2, 
                            lineType=cv2.LINE_4)

                # Get eyes coordinates
                right_eye, left_eye = tuple(face[4:6].astype(int)), tuple(face[6:8].astype(int))

                # Get nose tip coordinate
                nose = tuple(face[8:10].astype(int))

                # Get mouth coordinates
                left_mouth = (int(face[10]-10), int(face[11]-10))
                right_mouth = (int(face[12]+10), int(face[13]+10))
                
                # Draw eyes
                cv2.circle(img=frame, center=right_eye, radius=10, color=(255,255,255), thickness=1) 
                cv2.circle(img=frame, center=left_eye, radius=10, color=(255,255,255), thickness=1) 

                # Draw noise tip
                cv2.circle(img=frame, center=nose, radius=8, color=(0,255,255), thickness=1) 

                # Draw mouth
                cv2.rectangle(img=frame, pt1=left_mouth, pt2=right_mouth, color=(255, 0, 255), thickness=1)
                


    cv2.imshow("Face detector", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
    
cam.release()
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('./data/giraffe-Kenya.png', 0)
plt.imshow(img, 'grey')
```

### Sobel edge detector

```python
sobel_x = cv2.Sobel(src = img, 
                    ddepth = cv2.CV_64F,  # 64 bit floating point
                    dx = 1,               # X direction 
                    dy = 0,
                    ksize = 3)
plt.imshow(sobel_x, 'grey')
```

```python
sobel_y = cv2.Sobel(src = img, 
                    ddepth = cv2.CV_64F,  # 64 bit floating point
                    dx = 0,               # X direction 
                    dy = 1,
                    ksize = 3)
plt.imshow(sobel_y, 'grey')
```

```python
sobel_xy = cv2.magnitude(sobel_x, sobel_y)
plt.imshow(sobel_xy, 'grey')
```

### Laplacian

```python
laplacian = cv2.Laplacian(src=img, ddepth=cv2.CV_64F, ksize=3)
plt.imshow(laplacian, 'grey')
```

```python
laplacian_abs =cv2.convertScaleAbs(laplacian)
plt.imshow(laplacian_abs, 'grey')
```

### Canny

```python
canny_edges = cv2.Canny(image=img, threshold1=180, threshold2=200)
plt.imshow(canny_edges, 'grey')
```
